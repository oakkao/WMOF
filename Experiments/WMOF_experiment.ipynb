```python
# Calculate number of points in radius
@jit(nopython=True)
def _point_in_radius(dm, sort_idx, win_size, n):
  idx = np.ones((win_size,n), dtype = np.int64)
  for i in range(win_size):
    pre_distance = -1
    point_count = n
    for j in range(n-1, -1, -1):
      obv_point = sort_idx[i][j]
      if dm[i][obv_point] != pre_distance:
        point_count = j
        pre_distance = dm[i][obv_point]
      idx[i][obv_point] = point_count
  return idx
```

```python
# Calculate mass ratio matrix
@jit(nopython=True)
def _Var_Massratio(Data,window):
  # Beware of large numbers, it might overflow python int
    n = len(Data)
    mass = np.zeros(n)
    mass2 = np.zeros(n)
    assert(window > 0)

    # slicing window through data
    for start_point in range(0,n,window):
      stop_point = min(start_point+ window, n)
      Current_Data = Data[start_point : stop_point]
      win_size = stop_point - start_point

      with objmode(window_dm = "i8[:, :]", sort_idx = "i8[:, :]"):
        window_dm = cdist(Current_Data, Data)
        sort_idx = np.argsort(window_dm)

      # when pairwise distance is same, the index is also same
      current_idx = _point_in_radius(window_dm, sort_idx, win_size, n)

      # calculate all current points
      for i in range(start_point, stop_point):
        for j in range(i+1, stop_point):
          m = (current_idx[j%window][i]*1.0 + 1)/ (current_idx[i%window][j] + 1)
          mass[i] += m
          mass2[i] += m**2
          mass[j] += 1/m
          mass2[j] += 1/m**2

      # calculate remaining points
      for i in range(stop_point,n):
        with objmode(dm = "i8[:, :]", sort_remain = "i8[:, :]"):
          dm = cdist([Data[i]], Data)
          sort_remain = np.argsort(dm)

        # when pairwise distance is same, the index is also same
        idx = _point_in_radius(dm, sort_remain, 1, n)

        for j in range(start_point, stop_point):
          m = (current_idx[j%window][i]*1.0 + 1 )/ (idx[0][j] + 1)
          mass[i] += m
          mass2[i] += m**2
          mass[j] += 1/m
          mass2[j] += 1/m**2

    var = mass2/(n-1)-(mass/(n-1))**2
    return var
```

```python
# Calculate windowing mass ratio
@jit(nopython=True)
def _Window_Massratio(Data,Window,Overlap_ratio):
  n =Data.shape[0]
  scores = np.zeros(n)

  assert(Window > 0 )
  assert(0.0 <= Overlap_ratio <= 0.5)

  overlap_size = int(Overlap_ratio * Window)
  mid_overlap_point = int(Window - overlap_size/2)
  score_count = 0

  # Windowing data and send to MOF
  for start_point in range(0, n , Window - overlap_size):
    stop_point = min(n, start_point + Window)
    w = stop_point - start_point
    current_data = Data[start_point : stop_point]
    # Last data window size (w) <= Window
    current_scores = _Var_Massratio(current_data,w)

    # Assign score to data points
    stop_score_count = min(n, start_point + mid_overlap_point)
    for i in range(score_count, stop_score_count):
      scores[i] = current_scores[i - start_point]
      score_count += 1

  return scores
```

```python
# Calculate stream windowing mass ratio
@jit(nopython=True)
def _Stream_Window_Massratio(Data, Window, Overlap_ratio, Start_idx):
  n =Data.shape[0]
  scores = np.zeros(n)

  assert(Window > 0 )
  assert(0.0 <= Overlap_ratio <= 0.5)

  mid_overlap_point = int((1 - Overlap_ratio/2) * Window)

  scores = _Var_Massratio(Data,Window)

  return mid_overlap_point - int((1 - Overlap_ratio) * Window), Window - int((1 - Overlap_ratio) * Window), scores[Start_idx : mid_overlap_point] # start_idx, len, scores
```

## Class

```python
# from adbench.myutils import Utils
class WMOF:

# Windowing mass-ratio-variance based outlier factor (WMOF)
# This algorithm is an extension of the mass-ratio-variance outlier factor algorithm (MOF).
# WMOF operates on overlapping windows of fixed size, specified by the user.
# The use of overlapping windows ensures that anomalies occurring at window boundaries are not missed.
# For each window, the MOF score is computed for all data points within the window.

# Parameters-free
# ----------
  def __init__(self, Window=1000, Overlap_ratio = 0.2, model_name:str=None):
    self.name='WMOF'
    self.Data = []
    self.DataCount = 0 # for fit_score
    self.start_index = 0 # for fit_score
    self.window_size = Window # for fit_score
    self.overlap_ratio = Overlap_ratio # for fit_score
    self.decision_scores_ = np.array([], dtype = np.float64)
    self.Anomaly = np.array([])

    #for pyod
    # self.utils = Utils()

  def fit(self,X_train):

    '''
    Parameters
    ----------
    Data : numpy array of shape (n_samples, n_features)
        The input samples.
    Window : integer (int)
        number of points for each calculation
        default window size is 1000.
    Overlap_ratio : float
        0.0 <= Overlap_ratio <= 0.5
        A Overlap_ratio between window frame.
        default ratio is 0.2
    '''
    '''
    Returns
    -------
    self : object
    '''

  #  Fitted estimator.
    Data = X_train
    self.Data = Data

  # Calculate Windowing mass ratio variance (WMOF)
    #self.decision_scores_= _Window_Massratio(Data,Window,Overlap_ratio)
    return self

  def decision_function(self, X, Window =1000):
    X = np.array(X)
    self.decision_scores_= _Window_Massratio(X,Window,0.2)
    return self.decision_scores_

  def fit_score(self, x):
    self.Data.append(x)
    self.DataCount += 1
    score = []
    if self.DataCount == self.window_size:

      self.start_index, self.DataCount, score =  _Stream_Window_Massratio(np.array(self.Data), self.window_size, self.overlap_ratio, self.start_index)
      self.Data = self.Data[int((1 - self.overlap_ratio) * self.window_size):]

    return score

  def fit_last_score(self):
    self.start_index, self.DataCount, score =  _Stream_Window_Massratio(np.array(self.Data), len(self.Data), 0, self.start_index)
    self.Data = self.Data[int((1 - self.overlap_ratio) * self.window_size):]

    return score

  def detectAnomaly(self, theshold):
    '''
    Parameters
    ----------
    theshold : float
        A theshold value for detect anomaly points
    '''
    '''
    Returns
    -------
    idx : numpy array of shape (n_samples,)
        An index array of anomaly points in data
    '''
    # Check data avaliablity
    assert(len(self.Data) != 0)

    if self.decision_scores_.shape[0] == 0:
      self.fit(self.Data)
    idx = np.squeeze(np.argwhere(self.decision_scores_ > theshold))
    self.Anomaly = np.append(self.Anomaly,idx).astype(np.int32)
    self.Anomaly = np.unique(self.Anomaly)

    return idx

  def detectStream(self, scores, tau = None, n = 0.01):
    '''
    Parameters
    ----------
    scores : numpy array of shape (N,)
        decision scores for the current window.
    tau : float
        A threshold value for detecting anomaly points.
        default tau is None, the threshold is determined by 'n'.
    n : float or int
        If float (0.01 <= n <= 0.49), it is the percentage of anomaly data points
        to select (e.g., 0.01 means the top 1% of scores).
        If int, it is the exact number of anomaly data points to select.
        default n is 0.01
    '''
    '''
    Returns
    -------
    idx : numpy array of shape (N,)
        An index array of anomaly points in the current window of data.
    '''
    # Check data availability
    assert(len(self.Data) != 0)

    # Ensure decision scores are available (by fitting if necessary)
    if len(scores) == 0:
      return np.array([])

    num_data_points = scores.shape[0]

    # 1. Check tau first
    if tau is not None:
      threshold = tau

    # 2. If tau is None, check n (integer or float)
    elif type(n) == int:
      k = min(n, num_data_points) # Number of anomalies to pick
      if k == 0:
          threshold = np.max(scores) + 1
      else:
          threshold = np.percentile(scores, 100 * (num_data_points - k) / num_data_points)

    elif type(n) == float:
      percentile_cutoff = 100 * (1.0 - n)
      if 100 < percentile_cutoff or percentile_cutoff <  50: # Check for safety if n > 0.5
          percentile_cutoff = 100

      threshold = np.percentile(scores, percentile_cutoff)

    else:
      threshold = np.max(scores) + 1 # Threshold that excludes all points

    idx = np.squeeze(np.argwhere(scores > threshold))

    if idx.ndim == 0:
        idx = np.array([idx])

    return idx
```

```python
import numpy as np
from pysad.models.integrations import ReferenceWindowModel
from pysad.utils import ArrayStreamer
from pyod.models.iforest import IForest
from pyod.models.lof import LOF
from pyod.models.ocsvm import OCSVM
from pyod.models.knn import KNN
from tqdm import tqdm


window_size_baseline = 1000
sliding_size_baseline = 500

# --- IForest ---
initial_window_iforest = 100
model_iforest = ReferenceWindowModel(
    model_cls=IForest, window_size=window_size_baseline, sliding_size=sliding_size_baseline,
    initial_window_X=X_drift[:initial_window_iforest], n_estimators=100
)

scores_drift_iforest = []
iterator_iforest = ArrayStreamer(shuffle=False)
for X, y in tqdm(iterator_iforest.iter(X_drift, y_drift), desc="IForest"):
    model_iforest.fit_partial(X)
    score = model_iforest.score_partial(X)
    scores_drift_iforest.append(score)


# --- LOF ---
initial_window_lof = 100
model_lof = ReferenceWindowModel(
    model_cls=LOF, window_size=window_size_baseline, sliding_size=sliding_size_baseline,
    initial_window_X=X_drift[:initial_window_lof], n_neighbors=50
)

scores_drift_lof = []
iterator_lof = ArrayStreamer(shuffle=False)
for X, y in tqdm(iterator_lof.iter(X_drift, y_drift), desc="LOF"):
    model_lof.fit_partial(X)
    score = model_lof.score_partial(X)
    scores_drift_lof.append(score)


# --- OCSVM ---
initial_window_ocsvm = 100
model_ocsvm = ReferenceWindowModel(
    model_cls=OCSVM, window_size=window_size_baseline, sliding_size=sliding_size_baseline,
    initial_window_X=X_drift[:initial_window_ocsvm]
)

scores_drift_ocsvm = []
iterator_ocsvm = ArrayStreamer(shuffle=False)
for X, y in tqdm(iterator_ocsvm.iter(X_drift, y_drift), desc="OCSVM"):
    model_ocsvm.fit_partial(X)
    score = model_ocsvm.score_partial(X)
    scores_drift_ocsvm.append(score)


# --- KNN ---
initial_window_knn = 100
model_knn = ReferenceWindowModel(
    model_cls=KNN, window_size=window_size_baseline, sliding_size=sliding_size_baseline,
    initial_window_X=X_drift[:initial_window_knn], n_neighbors=50
)

scores_drift_knn = []
iterator_knn = ArrayStreamer(shuffle=False)
for X, y in tqdm(iterator_knn.iter(X_drift, y_drift), desc="KNN"):
    model_knn.fit_partial(X)
    score = model_knn.score_partial(X)
    scores_drift_knn.append(score)


scores_drift_iforest = np.array(scores_drift_iforest)
scores_drift_lof = np.array(scores_drift_lof)
scores_drift_ocsvm = np.array(scores_drift_ocsvm)
scores_drift_knn = np.array(scores_drift_knn)

print("Baseline model scores generated.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score

# --- Configuration ---
step_size = 10
y_true = synthetic_df['y'].values
drift_signal = synthetic_df['drift'].values

# Dictionary of scores to evaluate
all_model_scores = {
    'WMOF': scores_drift_wmof,
    'IForest': scores_drift_iforest,
    'LOF': scores_drift_lof,
    'OCSVM': scores_drift_ocsvm,
    'KNN': scores_drift_knn
}

# 1. Performance Calculation
indices = []
performance_curves = {name: [] for name in all_model_scores.keys()}

for i in range(100, len(y_true), step_size):
    y_slice = y_true[:i]
    if len(np.unique(y_slice)) > 1:
        indices.append(i)
        for name, scores in all_model_scores.items():
            score_slice = scores[:i]
            auc = roc_auc_score(y_slice, score_slice)
            performance_curves[name].append(auc)

# 2. Plotting
# Increase global font size for high DPI
plt.rcParams.update({'font.size': 16})
plt.figure(figsize=(14, 7), dpi=500)

# Highlight Drift Regions
plt.fill_between(range(len(drift_signal)), 0, 1.1,
                 where=(drift_signal == 1),
                 color='red', alpha=0.15, label='Drift Region', zorder=1)

# Plot each model curve
colors = ['#1D9E75', '#2c3e50', '#e67e22', '#8e44ad', '#2980b9']
for (name, curve), color in zip(performance_curves.items(), colors):
    plt.plot(indices, curve, label=name, color=color, lw=2, zorder=3)

plt.xlabel('Index', fontsize=16)
plt.ylabel('Cumulative AUROC', fontsize=16)
plt.title('Performance of WMOF Relative to Baseline Models', fontsize=18)
plt.ylim(0.4, 1.05)
plt.xlim(0, len(y_true))
plt.grid(True, linestyle=':', alpha=0.4, zorder=2)
plt.legend(loc='lower left', frameon=True, shadow=True, ncol=2, fontsize=14)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

def generate_base(n_samples, mode='trend'):
    time = np.arange(n_samples)
    if mode == 'trend':
        return 50 + 0.01 * time
    elif mode == 'seasonal':
        return 50 + 10 * np.sin(time / 50)
    elif mode == 'both':
        return 50 + 0.01 * time + 10 * np.sin(time / 50)
    return np.full(n_samples, 50.0)

def add_anomalies(data, labels, pattern='single', decay_rate=3.0, decay_length=80):
    n_samples = len(data)
    if pattern == 'single':
        for _ in range(5):
            idx = np.random.randint(0, n_samples)
            data[idx] += 30
            labels[idx] = 1
    elif pattern == 'multi':
        for _ in range(5):
            start = np.random.randint(0, n_samples - 20)
            duration = 15
            data[start:start+duration] += 25
            labels[start:start+duration] = 1
    elif pattern == 'decay':
        for _ in range(3):
            # Ensure we have enough room for the spike (5) + decay length
            start = np.random.randint(0, n_samples - (5 + decay_length))
            data[start:start+5] -= 30
            labels[start:start+5] = 1
            # Parameterized decay effect over decay_length points
            decay = 24 * np.exp(-np.linspace(0, decay_rate, decay_length))
            data[start+5:start+5+decay_length] -= decay
    return data, labels

synthetic_datasets = []
base_modes = ['trend', 'seasonal', 'both']
anomaly_patterns = ['single', 'multi', 'decay']

for b_mode in base_modes:
    for a_pattern in anomaly_patterns:
        n = 5000
        data = generate_base(n, mode=b_mode) + np.random.normal(0, 1.0, n)
        labels = np.zeros(n, dtype=int)
        # Using 80 points for decay length as requested
        data, labels = add_anomalies(data, labels, pattern=a_pattern, decay_rate=2.0, decay_length=300)

        scaler = MinMaxScaler()
        data_scaled = scaler.fit_transform(data.reshape(-1, 1))
        time_feat = np.linspace(0, 1, n).reshape(-1, 1)
        X_all = np.concatenate([data_scaled, time_feat], axis=1)

        synthetic_datasets.append({
            "name": f"{b_mode}_{a_pattern}",
            "X": X_all,
            "y": labels,
            "description": f"Base: {b_mode}, Anomaly: {a_pattern}"
        })

# Add 10th dataset: Combined patterns with adjustable decay length
n_10 = 10000
data_10 = generate_base(n_10, mode='both') + np.random.normal(0, 1.0, n_10)
labels_10 = np.zeros(n_10, dtype=int)

data_10, labels_10 = add_anomalies(data_10, labels_10, pattern='single')
data_10, labels_10 = add_anomalies(data_10, labels_10, pattern='multi')
data_10, labels_10 = add_anomalies(data_10, labels_10, pattern='decay', decay_rate=3, decay_length=500)

scaler_10 = MinMaxScaler()
data_scaled_10 = scaler_10.fit_transform(data_10.reshape(-1, 1))
time_feat_10 = np.linspace(0, 1, n_10).reshape(-1, 1)
# X_all_10 = np.concatenate([data_scaled_10, time_feat_10], axis=1)

X_all_10 = np.concatenate([data_10.reshape(-1, 1), time_feat_10], axis=1)

synthetic_datasets.append({
    "name": "both_combined_all",
    "X": X_all_10,
    "y": labels_10,
    "description": "Base: both, Anomaly: combined (single + multi + decay)"
})
```

```python
def create_time_delay_embedding(X, y, delay=10):
    """
    Creates a Time-Delay Embedding.
    Each row will contain the current value and the previous (delay-1) values.
    """
    n_samples, n_features = X.shape
    # We only embed the first column (the signal).
    # The second column is usually normalized time, which we can keep as a separate feature.
    signal = X[:, 0]

    embedded_features = []
    for i in range(delay):
        # Shift the signal and pad with the first value to maintain length
        shifted = np.roll(signal, i)
        shifted[:i] = signal[0]
        embedded_features.append(shifted)

    # Stack into [n_samples, delay] and reverse so index 0 is current, index 1 is t-1, etc.
    X_embedded = np.column_stack(embedded_features)

    # Optionally append the normalized time feature if it exists
    if n_features > 1:
        X_final = np.column_stack([X_embedded, X[:, 1]])
    else:
        X_final = X_embedded

    return X_final, y

embedded_synthetic_datasets = []

for dataset in synthetic_datasets:
    X_tde, y_tde = create_time_delay_embedding(dataset['X'], dataset['y'], delay=10)

    embedded_synthetic_datasets.append({
        "name": f"{dataset['name']}_TDE10",
        "X": X_tde,
        "y": y_tde,
        "description": f"{dataset['description']} with 10-step Time-Delay Embedding"
    })

print(f"Created {len(embedded_synthetic_datasets)} Time-Delay Embedded datasets.")
print(f"New feature dimension: {embedded_synthetic_datasets[0]['X'].shape[1]}")
```

```python
from sklearn.metrics import roc_auc_score

results = []

# synthetic_datasets

for dataset in embedded_synthetic_datasets:
    print(f"\nProcessing {dataset['name']}...")
    model = WMOF(Window=1000, Overlap_ratio=0.1)
    iterator = ArrayStreamer(shuffle=False)
    preprocessor = InstanceUnitNormScaler()

    all_scores = []
    for X, _ in tqdm(iterator.iter(dataset['X'], dataset['y']), desc=dataset['name']):
        X_p = preprocessor.fit_transform_partial(X.reshape(1, -1))
        score_batch = model.fit_score(X)
        if len(score_batch) > 0:
            all_scores.extend(score_batch)

    # Get remaining scores
    last_batch = model.fit_last_score()
    if len(last_batch) > 0:
        all_scores.extend(last_batch)

    # Align lengths
    final_scores = np.array(all_scores)[:len(dataset['y'])]

    # Calculate Quantitative Metric
    auc = roc_auc_score(dataset['y'], final_scores)
    results.append({"Dataset": dataset['name'], "AUROC": auc})

    # Visualization
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 6))
    ax1.plot(dataset['X'][:, 0], alpha=0.5, label='Signal')
    ax1.scatter(np.where(dataset['y']==1)[0], dataset['X'][dataset['y']==1, 0], color='red', s=5, label='Anomaly')
    ax1.set_title(f"Data: {dataset['name']} | True Labels")
    ax1.legend(loc='upper right')

    ax2.plot(final_scores, color='green', label='WMOF Score')
    ax2.set_title(f"WMOF Scores (AUROC: {auc:.4f})")
    ax2.legend(loc='upper right')
    plt.tight_layout()
    plt.show()

# Display Performance Summary
summary_df = pd.DataFrame(results)
print("\nQuantitative Summary (WMOF 3x3 Grid + Combined):")
print(summary_df.to_markdown(index=False))
```
